## Evaluation Insights

### Head-to-Head Accuracy vs. Actual Result
- **MRP MAE vs Actual**: 17.30 pp
- **PredictIt MAE vs Actual**: 4.31 pp
- **Mean Divergence (MRP vs Market)**: 18.35 pp

### Analysis
In 2024, **prediction markets were significantly more accurate than the MRP model**. While the market prices tracked the final outcome quite closely (4.31 pp error), the MRP model struggled immensely with a high 17.30 pp error.

The large 18.35 pp divergence highlights the 'data density gap' created by the mid-campaign transition from Biden to Harris. While traders in the market were able to incorporate this information instantly into their pricing, the MRP model, which processes rolling windows of Twitter data, was slower to adapt. This emphasizes a current limitation of the automated pipeline: its sensitivity to abrupt structural changes in the candidate set which can temporarily invalidate historical bias patterns.
