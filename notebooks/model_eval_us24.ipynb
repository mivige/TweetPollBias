This notebook runs the Sample-Calibrated Weighted GLM (SCWG) pipeline for the 2024 US Presidential Election and compares the results against actual outcomes and PredictIt prediction markets.

```python
from bias_analysis.modeling.train import main
from bias_analysis.plots import generate_adjusted_dashboard

ELECTION = "us24"

print(f"--- Running SCWG Pipeline for {ELECTION} ---")
main(election=ELECTION)

print(f"\n--- Generating Dashboard and Metrics for {ELECTION} ---")
generate_adjusted_dashboard(election=ELECTION)
```

## Evaluation Insights

### Head-to-Head Accuracy vs. Actual Result
- **SCWG MAE vs Actual**: 16.74 pp
- **PredictIt MAE vs Actual**: 4.06 pp
- **Mean Divergence (MRP vs Market)**: 18.66 pp

### Analysis
In 2024, **prediction markets were significantly more accurate than the SCWG model**. While the market prices tracked the final outcome quite closely (4.06 pp error), the model struggled immensely with a high error.

The large 18.66 pp divergence highlights the 'data density gap' created by the mid-campaign transition from Biden to Harris. While traders in the market were able to incorporate this information instantly into their pricing, the model, which processes rolling windows of Twitter data, was slower to adapt. This emphasizes a current limitation of the automated pipeline: its sensitivity to abrupt structural changes in the candidate set which can temporarily invalidate historical bias patterns.
