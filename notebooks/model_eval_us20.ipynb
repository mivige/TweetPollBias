### Analysis
In 2020, the **prediction market slightly outperformed the MRP model** in terms of overall proximity to the final vote share. While the MRP model performed competitively (6.96 pp error), the market prices remained closer to the actual result on average (4.11 pp error).

The 9.42 pp divergence between the two signals shows that they captured different dynamics. The market's superior performance suggests it accurately priced in certain regional or late-stage trends that the demographic-based MRP model, relying solely on Twitter data, did not fully resolve.
